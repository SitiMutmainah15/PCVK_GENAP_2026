```python
import numpy as np
import matplotlib.pyplot as plt


# membuat canvas hitam
black_int16 = np.zeros((512,512,3), dtype=np.int16)

black_int32 = np.zeros((512,512,3), dtype=np.int32)


# melihat informasi data

print("INT16")
print("dtype :", black_int16.dtype)
print("min   :", black_int16.min())
print("max   :", black_int16.max())
print("nbytes:", black_int16.nbytes)


print("\nINT32")
print("dtype :", black_int32.dtype)
print("min   :", black_int32.min())
print("max   :", black_int32.max())
print("nbytes:", black_int32.nbytes)


# menampilkan gambar

plt.figure(figsize=(10,5))

plt.subplot(1,2,1)
plt.imshow(black_int16)
plt.title("Black Image INT16")
plt.axis("off")


plt.subplot(1,2,2)
plt.imshow(black_int32)
plt.title("Black Image INT32")
plt.axis("off")


plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt
from skimage import io


# membaca gambar dengan skimage
img_skimage = io.imread("KTM IIN2.jpeg")


# membaca gambar dengan OpenCV
img_cv = cv2.imread("KTM IIN2.jpeg")


# konversi BGR ke RGB
img_rgb = cv2.cvtColor(img_cv, cv2.COLOR_BGR2RGB)



plt.figure(figsize=(12,5))


plt.subplot(1,2,1)
plt.imshow(img_skimage)
plt.title("Skimage RGB")
plt.axis("off")


plt.subplot(1,2,2)
plt.imshow(img_rgb)
plt.title("OpenCV BGR ke RGB")
plt.axis("off")


plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt

img = cv2.imread("KTM IIN2.jpeg")

# ubah BGR ke RGB
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)


# Membuat kombinasi Merah-Biru
red_blue = img_rgb.copy()
red_blue[:,:,1] = 0   # channel hijau dibuat 0


# Membuat kombinasi Hijau-Biru
green_blue = img_rgb.copy()
green_blue[:,:,0] = 0   # channel merah dibuat 0


# Menampilkan hasil
plt.figure(figsize=(10,5))


plt.subplot(1,2,1)
plt.imshow(red_blue)
plt.title("Kombinasi Merah-Biru")
plt.axis("off")


plt.subplot(1,2,2)
plt.imshow(green_blue)
plt.title("Kombinasi Hijau-Biru")
plt.axis("off")


plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt


# membaca gambar
img = cv2.imread("KTM IIN2.jpeg")

# ubah warna BGR ke RGB
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)


# pencerminan horizontal
flip_horizontal = cv2.flip(img_rgb, 1)


# pencerminan vertikal
flip_vertical = cv2.flip(img_rgb, 0)


# pencerminan horizontal dan vertikal
flip_both = cv2.flip(img_rgb, -1)



# menampilkan hasil

plt.figure(figsize=(12,5))


plt.subplot(1,3,1)
plt.imshow(flip_vertical)
plt.title("Pencerminan Vertikal")
plt.axis("off")


plt.subplot(1,3,2)
plt.imshow(flip_horizontal)
plt.title("Pencerminan Horizontal")
plt.axis("off")


plt.subplot(1,3,3)
plt.imshow(flip_both)
plt.title("Vertikal dan Horizontal")
plt.axis("off")


plt.show()
```
