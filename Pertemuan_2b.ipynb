```python
import numpy as np
import cv2 as cv
import matplotlib.pyplot as plt

# membuat canvas hitam
black_img = np.zeros(shape=(512,512,3), dtype=np.uint8)

# kotak hijau
cv.rectangle(
    black_img,
    pt1=(384,0),
    pt2=(510,150),
    color=(0,255,0),
    thickness=10
)

# kotak biru
cv.rectangle(
    black_img,
    pt1=(200,200),
    pt2=(300,300),
    color=(255,0,0),
    thickness=15
)

# lingkaran merah penuh
cv.circle(
    black_img,
    center=(400,400),
    radius=50,
    color=(0,0,255),
    thickness=-1
)

# lingkaran merah outline
cv.circle(
    black_img,
    center=(100,100),
    radius=50,
    color=(0,0,255),
    thickness=8
)

# garis diagonal putih
cv.line(
    black_img,
    pt1=(0,0),
    pt2=(511,511),
    color=(255,255,255),
    thickness=5
)

# menambahkan teks
font = cv.FONT_HERSHEY_SIMPLEX

cv.putText(
    black_img,
    text='Hello',
    org=(10,500),
    fontFace=font,
    fontScale=4,
    color=(255,255,0),
    thickness=2,
    lineType=cv.LINE_AA
)

# tampilkan hasil
plt.imshow(cv.cvtColor(black_img, cv.COLOR_RGB2BGR))
plt.show()
```

```python
img = cv2.imread("KTM CONTOH.png")

img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# mengambil channel RGB
R = img_rgb[:,:,0]
G = img_rgb[:,:,1]
B = img_rgb[:,:,2]
# Red-Blue
RB = img_rgb.copy()
RB[:,:,1] = 0

plt.figure(figsize=(10,5))

plt.subplot(1,2,1)
plt.imshow(RB)
plt.title("Channel Red-Blue")
plt.axis("off")
# Green-Blue
GB = img_rgb.copy()
GB[:,:,0] = 0

plt.subplot(1,2,2)
plt.imshow(GB)
plt.title("Channel Green-Blue")
plt.axis("off")

plt.show()
```
