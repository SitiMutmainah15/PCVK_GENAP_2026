```python
import numpy as np
import cv2 as cv
import matplotlib.pyplot as plt

# membuat canvas hitam
black_img = np.zeros(shape=(512,512,3), dtype=np.uint8)

# kotak hijau
cv.rectangle(
    black_img,
    pt1=(384,0),
    pt2=(510,150),
    color=(0,255,0),
    thickness=10
)

# kotak biru
cv.rectangle(
    black_img,
    pt1=(200,200),
    pt2=(300,300),
    color=(255,0,0),
    thickness=15
)

# lingkaran merah penuh
cv.circle(
    black_img,
    center=(400,400),
    radius=50,
    color=(0,0,255),
    thickness=-1
)

# lingkaran merah outline
cv.circle(
    black_img,
    center=(100,100),
    radius=50,
    color=(0,0,255),
    thickness=8
)

# garis diagonal putih
cv.line(
    black_img,
    pt1=(0,0),
    pt2=(511,511),
    color=(255,255,255),
    thickness=5
)

# menambahkan teks
font = cv.FONT_HERSHEY_SIMPLEX

cv.putText(
    black_img,
    text='Hello',
    org=(10,500),
    fontFace=font,
    fontScale=4,
    color=(255,255,0),
    thickness=2,
    lineType=cv.LINE_AA
)

# tampilkan hasil
plt.imshow(cv.cvtColor(black_img, cv.COLOR_RGB2BGR))
plt.show()
```

```python
img = cv2.imread("KTM CONTOH.png")

img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# mengambil channel RGB
R = img_rgb[:,:,0]
G = img_rgb[:,:,1]
B = img_rgb[:,:,2]
# Red-Blue
RB = img_rgb.copy()
RB[:,:,1] = 0

plt.figure(figsize=(10,5))

plt.subplot(1,2,1)
plt.imshow(RB)
plt.title("Channel Red-Blue")
plt.axis("off")
# Green-Blue
GB = img_rgb.copy()
GB[:,:,0] = 0

plt.subplot(1,2,2)
plt.imshow(GB)
plt.title("Channel Green-Blue")
plt.axis("off")

plt.show()
```

```python
img = cv2.imread("KTM CONTOH.png")
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
img_box = img_rgb.copy()
tinggi, lebar, channel = img_box.shape

# membuat 5 kotak random
for i in range(5):

    x1 = np.random.randint(0, lebar-100)
    y1 = np.random.randint(0, tinggi-100)

    ukuran = np.random.randint(30,100)

    x2 = x1 + ukuran
    y2 = y1 + ukuran

    warna = (
        np.random.randint(0,255),
        np.random.randint(0,255),
        np.random.randint(0,255)
    )

    cv2.rectangle(
        img_box,
        (x1,y1),
        (x2,y2),
        warna,
        thickness=5
    )
plt.figure(figsize=(4,5))
plt.imshow(img_box)
plt.title("5 Kotak Random pada Image KTM")
plt.axis("off")

plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt

img = cv2.imread("IIN.jpeg")
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
img_edit = img_rgb.copy()

# membuat rectangle pada wajah
cv2.rectangle(
    img_edit,
    pt1=(300,500),
    pt2=(520,750),
    color=(255,0,0),
    thickness=5
)
# membuat circle pada wajah
cv2.circle(
    img_edit,
    center=(410,620),
    radius=80,
    color=(0,255,0),
    thickness=5
)
plt.figure(figsize=(8,10))
plt.imshow(img_edit)
plt.title("Rectangle dan Circle pada Wajah")
plt.axis("off")
plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt

img = cv2.imread("kitten.jpg")
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

channel_B = img_rgb[:,:,2]
img_B = img_rgb.copy()

# membuat rectangle sudut bawah kiri
cv2.rectangle(
    img_B,
    pt1=(0, img_B.shape[0]-100),
    pt2=(100, img_B.shape[0]),
    color=(0,0,255),
    thickness=-1
)

plt.imshow(img_B)
plt.title("Rectangle pada Channel B RGB")
plt.axis("off")
plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt

img_text = img_B.copy()

font = cv2.FONT_HERSHEY_SIMPLEX

cv2.putText(
    img_text,
    text="kitten.jpg",
    org=(20,50),
    fontFace=font,
    fontScale=1.5,
    color=(255,255,0),
    thickness=3,
    lineType=cv2.LINE_AA
)

plt.figure(figsize=(8,8))
plt.imshow(img_text)
plt.title("Penambahan Nama File")
plt.axis("off")
plt.show()
```
