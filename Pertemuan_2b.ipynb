```python
import cv2 as cv
import matplotlib.pyplot as plt
from google.colab.patches import cv2_imshow

img = cv2.imread("KTM CONTOH.png")
img2 = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
plt.imshow(img2)
```

```python
img_gray = cv2.imread("KTM CONTOH.png", cv2.IMREAD_GRAYSCALE)

#TANPA CMAP
plt.subplot(1,2,1)
plt.imshow(img_gray)
plt.title("Tanpa CMAP")

#DENGAN CMAP
plt.subplot(1,2,2)
plt.imshow(img_gray, cmap="gray")
plt.title("Dengan CMAP")
plt.show()
```
