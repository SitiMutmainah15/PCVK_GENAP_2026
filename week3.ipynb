```python
import cv2
import numpy as np
from google.colab.patches import cv2_imshow


print("Melakukan inverse citra")
print("-----------------------")


# membaca gambar dari Drive
original = cv2.imread(
    '/content/drive/MyDrive/PCVK/Images/peppers.jpg'
)


# membuat tempat penyimpanan hasil inverse
inverse_image = np.zeros(
    original.shape,
    original.dtype
)


# akses per piksel
for i in range(original.shape[0]):
    for x in range(original.shape[1]):
        for c in range(original.shape[2]):

            # rumus inverse citra
            inverse_image[i,x,c] = 255 - original[i,x,c]


# menggabungkan gambar asli dan hasil inverse
final_frame = cv2.hconcat(
    (original, inverse_image)
)


# mengecilkan tampilan
final_frame = cv2.resize(
    final_frame,
    (800, 400)
)


# menampilkan hasil di Google Colab
cv2_imshow(final_frame)
```

```python
import cv2
import numpy as np
from google.colab.patches import cv2_imshow


print("Mengubah kontras dan tingkat kecerahan citra")
print("--------------------------------------------")


brightness = int(input("Masukkan tingkat kecerahan: "))
contrast = float(input("Masukkan kontras : "))


# membaca gambar
original = cv2.imread(
    '/content/drive/MyDrive/PCVK/Images/sunset.jpg'
)


# membuat tempat hasil
result = np.zeros(
    original.shape,
    original.dtype
)


# akses per piksel
for i in range(original.shape[0]):
    for j in range(original.shape[1]):
        for c in range(original.shape[2]):

            pixel = float(original[i,j,c])

            # rumus contrast + brightness
            result[i,j,c] = np.clip(
                contrast * pixel + brightness,
                0,
                255
            )


# menggabungkan gambar asli dan hasil
final_frame = cv2.hconcat(
    (original, result)
)


# mengecilkan tampilan
final_frame = cv2.resize(
    final_frame,
    (800, 400)
)


# menampilkan hasil
cv2_imshow(final_frame)
```

```python
import cv2
import numpy as np
from google.colab.patches import cv2_imshow


print("Mengubah tingkat kecerahan citra dengan Transformasi Log")
print("--------------------------------------------------------")


brightness = int(input("Masukkan nilai kecerahan: "))


# membaca gambar
original = cv2.imread(
    '/content/drive/MyDrive/PCVK/Images/houses.jpg'
)


# tempat hasil
log_image = np.zeros(
    original.shape,
    original.dtype
)


# gunakan nilai brightness sebagai konstanta c
c = brightness


# transformasi log
for i in range(original.shape[0]):
    for j in range(original.shape[1]):
        for k in range(original.shape[2]):

            log_image[i,j,k] = np.clip(
                c * np.log(1 + original[i,j,k]),
                0,
                255
            )


# gabungkan gambar
final_frame = cv2.hconcat(
    (original, log_image)
)


# kecilkan tampilan
final_frame = cv2.resize(
    final_frame,
    (800,400)
)


cv2_imshow(final_frame)
```

```python
import cv2
import numpy as np
from google.colab.patches import cv2_imshow


print("Transformasi Grayscale")
print("----------------------")


# membaca gambar
original = cv2.imread(
    '/content/drive/MyDrive/PCVK/Images/peppers.jpg'
)


# membuat tempat hasil
averaging = np.zeros(
    original.shape,
    original.dtype
)

lightness = np.zeros(
    original.shape,
    original.dtype
)

luminance = np.zeros(
    original.shape,
    original.dtype
)


# proses per piksel
for i in range(original.shape[0]):
    for j in range(original.shape[1]):

        # ambil nilai BGR
        b = int(original[i,j,0])
        g = int(original[i,j,1])
        r = int(original[i,j,2])


        # Averaging
        avg = (int(r)+int(g)+int(b)) / 3


        # Lightness
        light = (max(r,g,b) + min(r,g,b)) / 2


        # Luminance
        lum = (
            0.299*r +
            0.587*g +
            0.114*b
        )


        # masukkan ke semua channel
        averaging[i,j] = avg
        lightness[i,j] = light
        luminance[i,j] = lum



# gabungkan hasil
result1 = cv2.hconcat(
    (original, averaging)
)

result2 = cv2.hconcat(
    (original, lightness)
)

result3 = cv2.hconcat(
    (original, luminance)
)


# kecilkan tampilan
result1 = cv2.resize(result1,(800,400))
result2 = cv2.resize(result2,(800,400))
result3 = cv2.resize(result3,(800,400))


print("Averaging")
cv2_imshow(result1)


print("Lightness")
cv2_imshow(result2)


print("Luminance")
cv2_imshow(result3)
```

```python
import cv2
import numpy as np
from google.colab.patches import cv2_imshow


print("Menampilkan warna merah dan grayscale")
print("------------------------------------")


# membaca gambar
original = cv2.imread(
    '/content/drive/MyDrive/PCVK/Images/peppers.jpg'
)


# membuat tempat hasil
result = np.zeros(
    original.shape,
    original.dtype
)


# proses setiap piksel
for i in range(original.shape[0]):
    for j in range(original.shape[1]):

        # BGR OpenCV
        b = int(original[i,j,0])
        g = int(original[i,j,1])
        r = int(original[i,j,2])


        # mempertahankan warna merah saja
        if r > 120 and r > g * 1.5 and r > b * 1.5:

            result[i,j] = original[i,j]


        else:

            # grayscale averaging
            gray = (r + g + b) / 3

            result[i,j] = gray



# gabungkan gambar asli dan hasil
final_frame = cv2.hconcat(
    (original, result)
)


# kecilkan tampilan
final_frame = cv2.resize(
    final_frame,
    (800,400)
)


cv2_imshow(final_frame)
```

```python
import cv2
import numpy as np
from google.colab.patches import cv2_imshow


print("Gamma Correction pada citra")
print("---------------------------")


try:
    gamma = float(input("Masukkan nilai Gamma: "))

except ValueError:
    print("Error, not a number")


# membaca gambar
original = cv2.imread(
    '/content/drive/MyDrive/PCVK/Images/houses.jpg'
)


# membuat tempat hasil
gamma_image = np.zeros(
    original.shape,
    original.dtype
)


# konstanta
c = 255 / (255 ** gamma)


# proses setiap piksel
for i in range(original.shape[0]):
    for j in range(original.shape[1]):
        for k in range(original.shape[2]):

            gamma_image[i,j,k] = np.clip(
                c * (original[i,j,k] ** gamma),
                0,
                255
            )


# gabungkan gambar asli dan hasil
final_frame = cv2.hconcat(
    (original, gamma_image)
)


# kecilkan tampilan
final_frame = cv2.resize(
    final_frame,
    (800,400)
)


# tampilkan
cv2_imshow(final_frame)
```

```python
import cv2
import numpy as np
from google.colab.patches import cv2_imshow


print("Simulasi Image Depth")
print("--------------------")


# input bit depth
bit_depth = int(input("Masukkan bit depth (1-8): "))


# membaca gambar grayscale
original = cv2.imread(
    '/content/drive/MyDrive/PCVK/Images/peppers.jpg',
    cv2.IMREAD_GRAYSCALE
)


# membuat tempat hasil
depth_image = np.zeros(
    original.shape,
    original.dtype
)


# jumlah level grayscale
level = 255 / ((2 ** bit_depth) - 1)


# proses kuantisasi
for i in range(original.shape[0]):
    for j in range(original.shape[1]):

        pixel = original[i,j]


        depth_image[i,j] = np.round(
            pixel / level
        ) * level



# tampilkan informasi
print("Bit depth awal : 8 bit")
print("Bit depth baru :", bit_depth, "bit")
print("Jumlah level   :", 2 ** bit_depth)



# gabungkan gambar asli dan hasil
final_frame = cv2.hconcat(
    (original, depth_image)
)


# kecilkan tampilan
final_frame = cv2.resize(
    final_frame,
    (800,400)
)


# tampilkan
cv2_imshow(final_frame)
```

```python
import cv2
import numpy as np
import glob
from google.colab.patches import cv2_imshow


# gambar asli
original = cv2.imread(
    '/content/drive/MyDrive/PCVK/Images/galaxy.jpg'
)


# membaca semua gambar noise
cv_img = []

for img in glob.glob(
    '/content/drive/MyDrive/PCVK/Images/noises/*.jpg'
):
    n = cv2.imread(img)
    cv_img.append(n)



# fungsi PSNR
def hitung_psnr(original, hasil):

    mse = np.mean(
        (original.astype(float) - hasil.astype(float)) ** 2
    )

    if mse == 0:
        return 100

    psnr = 10 * np.log10(
        (255 ** 2) / mse
    )

    return psnr



jumlah_citra = [10,20,40,80,100]


for jumlah in jumlah_citra:

    # ambil jumlah citra tertentu
    images = cv_img[:jumlah]


    # averaging
    images_float = np.array(
        images,
        dtype=np.float32
    )


    average_image = np.mean(
        images_float,
        axis=0
    )


    average_image = np.uint8(
        average_image
    )


    # hitung PSNR setiap jumlah citra
    nilai_psnr = hitung_psnr(
        original,
        average_image
    )


    print(
        "Jumlah citra:",
        jumlah,
        " | PSNR:",
        nilai_psnr,
        "dB"
    )


    # tampilkan hasil
    cv2_imshow(average_image)
```

```python
import cv2
import numpy as np
from google.colab.patches import cv2_imshow


print("Gamma Correction")
print("-----------------")


# membaca gambar
original = cv2.imread(
    '/content/malam.jpg'
)


# nilai gamma
gamma = 0.6


# membuat tabel transformasi
table = np.array(
    [
        ((i / 255.0) ** gamma) * 255
        for i in np.arange(0,256)
    ]
).astype("uint8")


# menerapkan gamma correction
gamma_image = cv2.LUT(
    original,
    table
)


# gabungkan gambar asli dan hasil
final_frame = cv2.hconcat(
    (
        original,
        gamma_image
    )
)


final_frame = cv2.resize(
    final_frame,
    (800,400)
)


cv2_imshow(final_frame)
```
